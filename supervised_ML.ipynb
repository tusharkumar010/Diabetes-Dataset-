```python
# importing major liberacy
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

# importing warning remover
import warnings
warnings.filterwarnings('ignore')
```

```python
# Sklearm major liberies
from sklearn.model_selection import train_test_split
from sklearn.ensemble import StackingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import AdaBoostClassifier, GradientBoostingClassifier
from xgboost import XGBClassifier
```

# Imported major liberies from sklearn for training and testing algorithm

```python
df = pd.read_csv('diabetes.csv')
```

```python
df
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
x = df.drop('Outcome', axis=1)
y = df['Outcome']
```

```python
le = LabelEncoder()
y_encoded = le.fit_transform(y)
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y_encoded,test_size=0.2,random_state=42,stratify=y_encoded)
```

```python
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
base_learners = [
    ('dt', DecisionTreeClassifier(random_state=42)),
    ('svc', SVC(probability=True, kernel='rbf', random_state=42)),
    ('lr', LogisticRegression(max_iter=1000))
]
```

```python
meta_learner = LogisticRegression(max_iter=1000)
```

```python
stacking_clf = StackingClassifier(
    estimators=base_learners,
    final_estimator=meta_learner,
    cv=5
)
```

```python
stacking_clf.fit(x_train, y_train)
```

```python
y_pred = stacking_clf.predict(x_test)
```

```python
accuracy = accuracy_score(y_test,y_pred)
```

```python
accuracy
```

| Algorithm                        | Accuracy   |                                                                                                                                                                                                                                                                                            |
| -------------------------------- | ---------- | ----------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| **Stacking Classifier**          | **73.38%** |

#Description
An ensemble learning technique that combines predictions from multiple base models (**Decision Tree, SVM, and Logistic Regression**) and uses a meta-model (**Logistic Regression**) to make the final prediction. It leverages the strengths of different algorithms to improve overall performance. |
